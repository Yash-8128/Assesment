```python
class ClinicAppointment:
    def __init__(self):
        # Allowed doctors and time slots
        self.doctors = ["Dr. Amit", "Dr. Neha", "Dr. Arjun"]
        self.time_slots = ["10am", "11am", "12pm", "2pm", "3pm"]

        # Structure: appointments[doctor][slot] = list of patient dictionaries
        self.appointments = {
            doctor: {slot: [] for slot in self.time_slots}
            for doctor in self.doctors
        }

    # -----------------------------
    # BOOK APPOINTMENT
    # -----------------------------
    def book_appointment(self):
        print("\n--- Book Appointment ---")
        name = input("Enter patient name: ")
        age = input("Enter age: ")
        mobile = input("Enter mobile number: ")

        print("\nAvailable Doctors:")
        for i, doc in enumerate(self.doctors):
            print(f"{i+1}. {doc}")
        doc_choice = int(input("Choose doctor (1-3): "))
        doctor = self.doctors[doc_choice - 1]

        print("\nAvailable Time Slots:")
        for i, slot in enumerate(self.time_slots):
            print(f"{i+1}. {slot}")
        slot_choice = int(input("Choose time slot (1-5): "))
        slot = self.time_slots[slot_choice - 1]

        # Check max 3 appointments per doctor per slot
        if len(self.appointments[doctor][slot]) >= 3:
            print("\n❌ Slot is already full! Choose another time.\n")
            return

        # Add appointment
        self.appointments[doctor][slot].append({
            "name": name,
            "age": age,
            "mobile": mobile
        })

        print(f"\n✅ Appointment booked successfully with {doctor} at {slot}!\n")

    # -----------------------------
    # VIEW / CANCEL APPOINTMENT
    # -----------------------------
    def view_or_cancel(self):
        print("\n--- View/Cancel Appointment ---")
        mobile = input("Enter mobile number: ")

        found = False

        for doctor in self.doctors:
            for slot in self.time_slots:
                for appt in self.appointments[doctor][slot]:
                    if appt["mobile"] == mobile:
                        found = True
                        print(f"\n📌 Appointment Found:")
                        print(f"Name: {appt['name']}")
                        print(f"Age: {appt['age']}")
                        print(f"Doctor: {doctor}")
                        print(f"Time Slot: {slot}")

                        choice = input("\nDo you want to cancel? (yes/no): ").lower()
                        if choice == "yes":
                            self.appointments[doctor][slot].remove(appt)
                            print("\n🗑️ Appointment cancelled successfully.\n")
                        return

        if not found:
            print("\n❌ No appointment found for this mobile number.\n")

    # -----------------------------
    # SHOW APPOINTMENTS (OPTIONAL)
    # -----------------------------
    def show_all(self):
        print("\n--- All Appointments ---")
        for doctor in self.doctors:
            print(f"\nDoctor: {doctor}")
            for slot in self.time_slots:
                print(f"  {slot}: {len(self.appointments[doctor][slot])} bookings")


# ---------------------------------------------------
# RUN THE SYSTEM
# ---------------------------------------------------
clinic = ClinicAppointment()

while True:
    print("\n==== Clinic Appointment System ====")
    print("1. Book Appointment")
    print("2. View/Cancel Appointment")
    print("3. Show All Appointments")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        clinic.book_appointment()
    elif choice == '2':
        clinic.view_or_cancel()
    elif choice == '3':
        clinic.show_all()
    elif choice == '4':
        print("\nThank you! Exiting system.\n")
        break
    else:
        print("\nInvalid choice. Try again.\n")
```

```python
class SchoolManagement:
    def __init__(self):
        self.students = {}          # student_id → student details
        self.next_id = 1            # auto-generated ID starts from 1

    # ------------------------------------------------
    # NEW ADMISSION
    # ------------------------------------------------
    def new_admission(self):
        print("\n--- New Admission ---")
        name = input("Enter student name: ")
        age = int(input("Enter age: "))
        std_class = int(input("Enter class (1-12): "))
        mobile = input("Enter guardian's mobile number: ")

        # Validations
        if not (5 <= age <= 18):
            print("❌ Age must be between 5 and 18.\n")
            return

        if not (1 <= std_class <= 12):
            print("❌ Class must be between 1 and 12.\n")
            return

        if not (mobile.isdigit() and len(mobile) == 10):
            print("❌ Mobile number must be exactly 10 digits.\n")
            return

        # Assign unique ID
        student_id = self.next_id
        self.next_id += 1

        # Store record
        self.students[student_id] = {
            "name": name,
            "age": age,
            "class": std_class,
            "mobile": mobile
        }

        print(f"\n✅ Admission successful! Student ID = {student_id}\n")

    # ------------------------------------------------
    # VIEW STUDENT DETAILS
    # ------------------------------------------------
    def view_student(self):
        print("\n--- View Student Details ---")
        sid = int(input("Enter student ID: "))

        if sid not in self.students:
            print("❌ No student found with this ID.\n")
            return

        student = self.students[sid]
        print("\n📌 Student Details:")
        print(f"ID: {sid}")
        print(f"Name: {student['name']}")
        print(f"Age: {student['age']}")
        print(f"Class: {student['class']}")
        print(f"Guardian Mobile: {student['mobile']}\n")

    # ------------------------------------------------
    # UPDATE STUDENT INFO
    # ------------------------------------------------
    def update_student(self):
        print("\n--- Update Student Info ---")
        sid = int(input("Enter student ID: "))

        if sid not in self.students:
            print("❌ No student found with this ID.\n")
            return

        print("\nWhat do you want to update?")
        print("1. Mobile Number")
        print("2. Class")
        choice = input("Enter choice: ")

        if choice == "1":
            new_mobile = input("Enter new mobile number: ")
            if new_mobile.isdigit() and len(new_mobile) == 10:
                self.students[sid]["mobile"] = new_mobile
                print("✅ Mobile number updated.\n")
            else:
                print("❌ Invalid mobile number.\n")

        elif choice == "2":
            new_class = int(input("Enter new class (1–12): "))
            if 1 <= new_class <= 12:
                self.students[sid]["class"] = new_class
                print("✅ Class updated.\n")
            else:
                print("❌ Invalid class.\n")
        else:
            print("❌ Invalid choice.\n")

    # ------------------------------------------------
    # REMOVE STUDENT RECORD
    # ------------------------------------------------
    def remove_student(self):
        print("\n--- Remove Student Record ---")
        sid = int(input("Enter student ID: "))

        if sid in self.students:
            del self.students[sid]
            print("🗑️ Student record removed.\n")
        else:
            print("❌ No student found with this ID.\n")

    # ------------------------------------------------
    # RUN MENU
    # ------------------------------------------------
    def run(self):
        while True:
            print("\n===== School Management System =====")
            print("1. New Admission")
            print("2. View Student Details")
            print("3. Update Student Info")
            print("4. Remove Student Record")
            print("5. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":
                self.new_admission()
            elif choice == "2":
                self.view_student()
            elif choice == "3":
                self.update_student()
            elif choice == "4":
                self.remove_student()
            elif choice == "5":
                print("Exiting System... Goodbye!")
                break
            else:
                print("❌ Invalid choice. Try again.\n")


# START SYSTEM
school = SchoolManagement()
school.run()
```

```python
class BusReservation:
    def __init__(self):
        # Predefined routes with prices
        self.routes = {
            "Mumbai to Pune": 500,
            "Delhi to Jaipur": 600,
            "Bangalore to Mysore": 400,
            "Chennai to Coimbatore": 550,
            "Kolkata to Durgapur": 450
        }

        # ticket_id → ticket details
        self.tickets = {}

        # Auto-increment ticket number
        self.ticket_counter = 1

        # Route → seats booked (max seat = 40)
        self.seats = {route: [] for route in self.routes}

    # ----------------------------------------------------
    # SHOW AVAILABLE ROUTES
    # ----------------------------------------------------
    def show_routes(self):
        print("\n--- Available Bus Routes ---")
        for route, price in self.routes.items():
            print(f"{route} - ₹{price}")
        print()

    # ----------------------------------------------------
    # BOOK TICKET
    # ----------------------------------------------------
    def book_ticket(self):
        print("\n--- Book Ticket ---")
        name = input("Enter passenger name: ")
        age = int(input("Enter passenger age: "))
        mobile = input("Enter mobile number: ")

        # Mobile validation
        if not (mobile.isdigit() and len(mobile) == 10):
            print("❌ Mobile number must be 10 digits.\n")
            return

        print("\nChoose Route:")
        for i, route in enumerate(self.routes.keys()):
            print(f"{i+1}. {route}")

        route_choice = int(input("Enter choice number: "))
        route = list(self.routes.keys())[route_choice - 1]

        # Seat availability check
        if len(self.seats[route]) >= 40:
            print("\n❌ No seats available for this route.\n")
            return

        # Assign next seat
        seat_no = len(self.seats[route]) + 1
        self.seats[route].append(seat_no)

        # Generate unique ticket ID
        ticket_id = self.ticket_counter
        self.ticket_counter += 1

        # Save ticket details
        self.tickets[ticket_id] = {
            "name": name,
            "age": age,
            "mobile": mobile,
            "route": route,
            "seat": seat_no,
            "price": self.routes[route]
        }

        print(f"\n✅ Ticket Booked Successfully!")
        print(f"Your Ticket ID: {ticket_id}")
        print(f"Seat Number: {seat_no}\n")

    # ----------------------------------------------------
    # VIEW TICKET
    # ----------------------------------------------------
    def view_ticket(self):
        print("\n--- View Ticket ---")
        ticket_id = int(input("Enter Ticket ID: "))

        if ticket_id not in self.tickets:
            print("❌ Ticket not found.\n")
            return

        t = self.tickets[ticket_id]
        print("\n📌 Ticket Details:")
        print(f"Ticket ID: {ticket_id}")
        print(f"Name: {t['name']}")
        print(f"Age: {t['age']}")
        print(f"Mobile: {t['mobile']}")
        print(f"Route: {t['route']}")
        print(f"Seat No: {t['seat']}")
        print(f"Price: ₹{t['price']}\n")

    # ----------------------------------------------------
    # CANCEL TICKET
    # ----------------------------------------------------
    def cancel_ticket(self):
        print("\n--- Cancel Ticket ---")
        ticket_id = int(input("Enter Ticket ID: "))

        if ticket_id not in self.tickets:
            print("❌ Ticket not found.\n")
            return

        route = self.tickets[ticket_id]["route"]
        seat_no = self.tickets[ticket_id]["seat"]

        # Make seat available again
        self.seats[route].remove(seat_no)

        # Delete ticket
        del self.tickets[ticket_id]

        print("🗑️ Ticket cancelled successfully.\n")

    # ----------------------------------------------------
    # MENU
    # ----------------------------------------------------
    def run(self):
        while True:
            print("\n===== Bus Reservation System =====")
            print("1. Show Routes")
            print("2. Book Ticket")
            print("3. View Ticket")
            print("4. Cancel Ticket")
            print("5. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":
                self.show_routes()
            elif choice == "2":
                self.book_ticket()
            elif choice == "3":
                self.view_ticket()
            elif choice == "4":
                self.cancel_ticket()
            elif choice == "5":
                print("Exiting system... Goodbye!")
                break
            else:
                print("❌ Invalid choice. Try again.\n")


# Start system
bus = BusReservation()
bus.run()
```
